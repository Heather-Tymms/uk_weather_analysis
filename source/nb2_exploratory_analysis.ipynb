# Exploratory Analysis
This notebook will look into the initial analysis to see what data is held in the Met Office website. We will find the attributes that are collected/shown to the public and get a feel for the dataset.

```python
# packages
from clean import create_table
from location import Location
from location_attr import LocationAttributes
```

## Choose a location
First we will choose a location. If you would like to know what locations there are to choose from, either look at location.py or start writing "Location."and press "ctrl + space" which will show you a list of all weather stations around the UK.

```python
# location variables
LOCATION = Location.HEATHROW
location_attributes = LocationAttributes(LOCATION)

# data frame
location_attributes.df = create_table(
    location_name=LOCATION.value,
    location_url=location_attributes.url,
    data_refresh=False,
)
```

## Explore the data
Now we have a clean dataset from a chosen location, we can see what data is in it.

```python
# What are the attributes of this dataset
location_attributes.df.columns
```

## The columns
As we can see above, here are the columns in this dataset and what each of these represent.

- **year** - An integer that represents the year
- **decade** - An integer that represents the decade (analyst created in `clean.py`)
- **month** - An integer that represents the month
- **worded_month** - A string that represents the month (analyst created in `clean.py`)
- **tmax_degC** - A float (a number with decimals) that represents the maximum temperature that day in degrees celcius
- **tmin_degC** - A float (a number with decimals) that represents the minimum temperature that day in degrees celcius
- **af_days** - A float (a number with decimals) that represents the number of days of air frost
- **rain_mm** - A float that represents the amount of rain in the area in millimetres
- **sun_hours** - A float that represents the amount of sun in hours
- **is_predicted** - A boolean (True/False) to show whether the record is predictive or an observation.

### The attributes and the measurements
We can see that every month and year, we get measurements for temperature, rain, air frost and sun.
